```python
#  Task 2 : Cleaning and Structuring data 
```

```python
import json
```

```python
def clean_data(data):
    # remove users with missing values
    data["users"] = [user for user in data['users'] if user['name'].strip()]
    # to remove duplicate friends 
    for user in data["users"]:
        user['friends'] = list(set(user['friends']))
    # Remove inactive users 
    data['users'] = [user for user in data['users']if user['friends'] or user['liked_pages']]

    # Remove duplicate pages 
    unique = {}
    for page in data["pages"]:
        unique[page["id"]] = page
    data["pages"] = list(unique.values())
    return data 

# Load the Data 
data = json.load(open("data(CWH)2.json"))
data = clean_data(data)
json.dump(data , open("cleaned_data2.json" , "w"), indent = 4)
print("Data has beeen cleaned successfully")
```

```python
# Task 3 : People You May Know
```

```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        return json.load(f)

def people_you_may_know(user_id , data):
    user_friends = {}
    for user in data["users"]:
        user_friends[user["id"]] = set(user["friends"])
    if user_id not in user_friends:
        return []

    direct_friends = user_friends[user_id]
    suggestions = {}
    for friend in direct_friends :
        for mutual in user_friends[friend]:
            if mutual != user_id and mutual not in direct_friends:
                # count mutual friends
                suggestions[mutual]= suggestions.get(mutual , 0)+ 1
    sorted_suggestions = sorted(suggestions.items(), key=lambda x: x[1], reverse = True)
    return[user_id for user_id, _ in sorted_suggestions]

# load data
data = load_data("massive(CWH).json")
user_id = 3
recc = people_you_may_know(user_id , data)
print(recc)

```

```python
# Task 4 : Pages you might like 
```

```python
import json
```

```python
# load the data 
def load_data(filename):
    with open(filename , "r") as f :
        return json.load(f)

# function to write a user might like based on common interests 
def Pages_you_might_like(user_id , data):
    # dictonarily to store user interactions with pages 
    user_pages = {}
    # populate the dictornary
    for user in data['users']:
        user_pages[user["id"]]= set(user["liked_pages"])

    # if user is not foud return an emoty list
    if user_id not in user_pages:
       return []

    user_liked_pages = user_pages[user_id]
    page_suggestions = {}

    for other_user, pages in user_pages.items():
        if other_user != user_id:
            shared_pages = user_liked_pages.intersection(pages)
            for page in pages:
                if page not in user_liked_pages:
                    page_suggestions[page] = page_suggestions.get(page, 0) + len(shared_pages)
                    
    sorted_pages = sorted(page_suggestions.items(), key=lambda x: x[1], reverse= True)
    return [page_id for page_id, _ in sorted_pages]

data = load_data("massive(CWH).json")
user_id = 1
page_recommendations = Pages_you_might_like(user_id , data)
print(page_recommendations)
                                
```
